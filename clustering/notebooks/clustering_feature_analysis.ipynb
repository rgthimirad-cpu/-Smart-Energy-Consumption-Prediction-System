```python
cluster_features = [

    # Weather influence
    "Temperature",
    "Humidity",
    "Wind_Speed",

    # Main energy consumption
    "Zone_1_Power_Consumption",
    "Zone_2_Power_Consumption",
    "Zone_3_Power_Consumption",

    # Time behaviour
    "Hour",
    "Day_of_Week",
    "Month",
    "Is_Weekend",
    "Is_Peak_Hour",

    # Recent consumption behaviour
    "Zone_1_lag_1",
    "Zone_2_lag_1",
    "Zone_3_lag_1",

    # Daily consumption trends
    "Zone_1_roll_mean_24h",
    "Zone_2_roll_mean_24h",
    "Zone_3_roll_mean_24h"

]
```

## Selected Clustering Features

The selected features represent four main aspects of energy consumption behaviour:

1. Energy demand:
- Zone_1_Power_Consumption
- Zone_2_Power_Consumption
- Zone_3_Power_Consumption

2. Time-based behaviour:
- Hour
- Day_of_Week
- Month
- Is_Weekend
- Is_Peak_Hour

3. Environmental influence:
- Temperature
- Humidity
- Wind_Speed

4. Consumption trends:
- Lag features
- 24-hour rolling mean features

These features were selected because they describe energy usage patterns and allow clustering algorithms to identify similar consumption behaviours.
